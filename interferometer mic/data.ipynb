```python
import numpy as np
import scipy
import matplotlib.pyplot as plt
import moku
import moku.instruments
import moku_util as mu
```

```python
ip_address = "[fe80::7269:79ff:feb9:468e]"
logger = moku.instruments.Datalogger(ip_address, force_connect=True)
```

```python
mu.setup_two_channel_logger(logger)
```

```python
sampling_frequency = 44100.0      # samples per second
duration = 20.0                  # duration of acquisition in seconds
delay = 0                       # optional acquisition delay in seconds

# Change this to your desired save directory. The .li file will
# be saved in this directory with a name that includes your specified
# filename_prefix and a timestamp set by the DataLogger.
savedir = './data'

# Log data from the Moku:Go and download it locally to your computer.
filename = mu.log_data(
    logger,
    sampling_frequency,
    duration,
    delay=delay,
    filename_prefix="mokudata",
    save_directory=savedir,
    # use_trigger=True,
    # trigger_source="Input1",
    # trigger_level=0.25,
)
```

```python
logger.relinquish_ownership()
```

```python
# Load the data using the filename returned by the log_data function.
df = mu.DataFile(".data/mokudata.")
%matplotlib widget
fig, ax = df.plot_data(input=[0], microseconds=False, xlim=(0, 20))
```
